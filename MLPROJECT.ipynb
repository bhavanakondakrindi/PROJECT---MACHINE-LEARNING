```python
# Separate input features and traffic labels

X = df.drop(columns=["timestamp", "road_segment_id", "label"])
y = df["label"]

print("Number of features:", X.shape[1])
print("Number of observations:", X.shape[0])
```

```python
print("Features used for PCA and t-SNE:\n")

for i, feature in enumerate(X.columns, start=1):
    print(i, "→", feature)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
```

```python
import numpy as np

print("Mean of first 5 standardized features:")
print(X_scaled[:, :5].mean(axis=0))

print("\nStandard deviation of first 5 standardized features:")
print(X_scaled[:, :5].std(axis=0))
```

```python
from sklearn.decomposition import PCA

pca = PCA()

X_pca = pca.fit_transform(X_scaled)

print("Original dimensions:", X_scaled.shape)
print("PCA dimensions:", X_pca.shape)
```

```python
explained_variance = pca.explained_variance_ratio_

for i, variance in enumerate(explained_variance, start=1):
    print(f"PC{i}: {variance * 100:.2f}%")
```

```python
pca_2 = PCA(n_components=2)

X_pca_2 = pca_2.fit_transform(X_scaled)

explained_2d = pca_2.explained_variance_ratio_.sum()

print(f"Variance explained by PC1: {pca_2.explained_variance_ratio_[0] * 100:.2f}%")
print(f"Variance explained by PC2: {pca_2.explained_variance_ratio_[1] * 100:.2f}%")
print(f"Total variance explained by PC1 + PC2: {explained_2d * 100:.2f}%")
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

for label in y.unique():
    mask = y == label

    plt.scatter(
        X_pca_2[mask, 0],
        X_pca_2[mask, 1],
        label=label,
        alpha=0.5,
        s=8
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization of Traffic Flow Data")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

```python
pca_results = pd.DataFrame(
    X_pca_2,
    columns=["PC1", "PC2"]
)

pca_results["label"] = y.values

pca_results.head()
```

```python
# Create a representative sample for t-SNE

sample_size = 5000

tsne_df = df.groupby("label", group_keys=False).apply(
    lambda x: x.sample(
        n=round(sample_size * len(x) / len(df)),
        random_state=42
    )
)

print("t-SNE sample shape:", tsne_df.shape)
print("\nTraffic labels in t-SNE sample:")
print(tsne_df["label"].value_counts())
```

```python
X_tsne = tsne_df.drop(
    columns=["timestamp", "road_segment_id", "label"]
)

y_tsne = tsne_df["label"]

print("t-SNE feature shape:", X_tsne.shape)
print("t-SNE label shape:", y_tsne.shape)
```

```python
scaler_tsne = StandardScaler()

X_tsne_scaled = scaler_tsne.fit_transform(X_tsne)

print("Scaled t-SNE data shape:", X_tsne_scaled.shape)
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(
    n_components=2,
    perplexity=30,
    learning_rate="auto",
    init="pca",
    random_state=42
)

X_tsne_2d = tsne.fit_transform(X_tsne_scaled)

print("t-SNE output shape:", X_tsne_2d.shape)
```

```python
tsne_results = pd.DataFrame(
    X_tsne_2d,
    columns=["tSNE1", "tSNE2"]
)

tsne_results["label"] = y_tsne.values

tsne_results.head()
```

```python
plt.figure(figsize=(10, 7))

for label in tsne_results["label"].unique():
    mask = tsne_results["label"] == label

    plt.scatter(
        tsne_results.loc[mask, "tSNE1"],
        tsne_results.loc[mask, "tSNE2"],
        label=label,
        alpha=0.6,
        s=10
    )

plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")
plt.title("t-SNE Visualization of Traffic Flow Data")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```

```python
pca_sample = PCA(n_components=2)

X_pca_sample = pca_sample.fit_transform(X_tsne_scaled)

print(
    f"PCA variance explained by PC1 + PC2: "
    f"{pca_sample.explained_variance_ratio_.sum() * 100:.2f}%"
)
```

```python
plt.figure(figsize=(10, 7))

for label in y_tsne.unique():
    mask = y_tsne == label

    plt.scatter(
        X_pca_sample[mask, 0],
        X_pca_sample[mask, 1],
        label=label,
        alpha=0.6,
        s=10
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization Using the Same Sample as t-SNE")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```
